```python
np.random.seed(42)
```

```python
months = pd.date_range(start="2025-01-01", periods=12, freq="ME")
data = pd.DataFrame(
    {
        "Month": months,
        "Total_Credit_Sales": np.random.randint(500000, 800000, 12),
        "Accounts_Receivable": np.random.randint(120000, 250000, 12),
        "Cost_of_Goods_Sold": np.random.randint(300000, 500000, 12),
        "Accounts_Payable": np.random.randint(80000, 150000, 12),
        "Current_Assets": np.random.randint(600000, 900000, 12),
        "Current_Liabilities": np.random.randint(300000, 500000, 12),
        "Operating_Cash_Flow": np.random.randint(50000, 150000, 12),
    }
)

print(data)
```

```python
data["DSO"] = (data["Accounts_Receivable"]/ data["Total_Credit_Sales"]) * 30 
print(data)
    
```

```python
data["DPO"] = (data["Accounts_Payable"] / data ["Cost_of_Goods_Sold"]) * 30 
print(data)
```

```python
data["Current_Ratio"] = data["Current_Assets"] / data["Current_Liabilities"] 
print(data[["Month", "DSO", "DPO", "Current_Ratio"]].tail(3))
```

```python
avg_monthly_inflow = data["Total_Credit_Sales"].mean()
avg_monthly_outflow = data["Cost_of_Goods_Sold"].mean()
print(avg_monthly_inflow)
print(avg_monthly_outflow)
```

```python
starting_cash= 100000
```

```python
forecast_period = ["Month + 30 days", "Month + 60 days", "Month + 90 days"]
project_inflows = [
    avg_monthly_inflow * (30/ data["DSO"].iloc[1]),
    avg_monthly_inflow,
    avg_monthly_inflow,
]

project_outflows = [avg_monthly_outflow * 0.95, avg_monthly_outflow, avg_monthly_outflow * 1.05]

print(project_inflows)
print(project_outflows)
```

```python
import pandas as pd

forecast_df = pd.DataFrame({
    "Timeline": pd.Series(forecast_period),
    "Project_Inflows": pd.Series(project_inflows),
    "Project_Outflows": pd.Series(project_outflows),
})
```

```python
print(forecast_df)
```

```python
forecast_df["Net_cash_Flow"] = (
    forecast_df["Project_Inflows"] - forecast_df ["Project_Outflows"]
)

print(forecast_df)
```

```python
forecast_df["Ending_Balance"] = starting_cash + forecast_df["Net_cash_Flow"].cumsum()
print(forecast_df)
```

```python
def check_liquidity_health(row):
    if row["Current_Ratio"] < 1.2 or row ["DSO"] > 55: 
        return"CRITICAL: High Risk of Cash Crunch"
    elif row["Current_Ratio"] < 1.5 or row["DSO"] > 40:
        return "Warning Monitor Closely"
    else:
        return "Healthy Liquidity stability"


data["Health_Status"]= data.apply(check_liquidity_health, axis = 1)
print(data[["Month", "DSO", "Current_Ratio", "Health_Status"]].tail(3))
```
